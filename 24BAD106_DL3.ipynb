```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import seaborn as sns

from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

# Display student details
print("CNN IMAGE CLASSIFICATION")
print("Name    : SASTHA JEYASRI A")
print("Roll No : 24BAD106")
print("TensorFlow Version:", tf.__version__)

# Set the dataset paths
dataset_path = "/content/intel_dataset"
train_path = os.path.join(dataset_path, "train")
val_path = os.path.join(dataset_path, "val")

# Define the two classes
classes = ["cat", "dog"]

print("\nDataset Path:", dataset_path)
print("Training folders:", os.listdir(train_path))
print("Validation folders:", os.listdir(val_path))

# Count the number of images in each class
print("\nIMAGE COUNT")

for folder_name, folder_path in [("Training", train_path), ("Validation", val_path)]:
    print("\n" + folder_name)

    for class_name in classes:
        class_path = os.path.join(folder_path, class_name)

        files = [
            f for f in os.listdir(class_path)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ]

        print(class_name, ":", len(files))

# Store image paths and their corresponding labels
image_paths = []
image_labels = []

for label, class_name in enumerate(classes):
    class_path = os.path.join(train_path, class_name)

    for file_name in os.listdir(class_path):

        # Select only image files
        if file_name.lower().endswith((".jpg", ".jpeg", ".png")):

            image_paths.append(
                os.path.join(class_path, file_name)
            )

            image_labels.append(label)

# Convert lists into NumPy arrays
image_paths = np.array(image_paths)
image_labels = np.array(image_labels)

print("\nTotal images in train folder:", len(image_paths))
print("Total cats:", np.sum(image_labels == 0))
print("Total dogs:", np.sum(image_labels == 1))

# Split the data into training and testing sets
# Stratify ensures that both cat and dog are present in both sets
X_train, X_test, y_train, y_test = train_test_split(
    image_paths,
    image_labels,
    test_size=0.20,
    random_state=42,
    stratify=image_labels,
    shuffle=True
)

print("\nTRAIN / TEST SPLIT")
print("Training images:", len(X_train))
print("Testing images :", len(X_test))

# Display class distribution in training data
print("\nTraining class distribution:")
print("Cats:", np.sum(y_train == 0))
print("Dogs:", np.sum(y_train == 1))

# Display class distribution in testing data
print("\nTesting class distribution:")
print("Cats:", np.sum(y_test == 0))
print("Dogs:", np.sum(y_test == 1))

# Set image size and batch size
IMG_SIZE = (128, 128)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE

# Function to load and preprocess images
def load_image(path, label):

    # Read the image file
    img = tf.io.read_file(path)

    # Decode the image into RGB format
    img = tf.image.decode_image(
        img,
        channels=3,
        expand_animations=False
    )

    # Resize the image to 128 x 128
    img = tf.image.resize(img, IMG_SIZE)

    # Convert image values to floating point
    img = tf.cast(img, tf.float32)

    return img, label

# Create the training dataset
train_ds = tf.data.Dataset.from_tensor_slices(
    (X_train, y_train)
)

# Load and preprocess training images
train_ds = train_ds.map(
    load_image,
    num_parallel_calls=AUTOTUNE
)

# Shuffle training images
train_ds = train_ds.shuffle(
    buffer_size=len(X_train),
    seed=42
)

# Divide images into batches
train_ds = train_ds.batch(BATCH_SIZE)

# Improve data loading speed
train_ds = train_ds.prefetch(AUTOTUNE)

# Create the testing dataset
test_ds = tf.data.Dataset.from_tensor_slices(
    (X_test, y_test)
)

# Load and preprocess testing images
test_ds = test_ds.map(
    load_image,
    num_parallel_calls=AUTOTUNE
)

# Divide test images into batches
test_ds = test_ds.batch(BATCH_SIZE)

# Improve data loading speed
test_ds = test_ds.prefetch(AUTOTUNE)

# Load the validation dataset from the val folder
val_ds = tf.keras.utils.image_dataset_from_directory(
    val_path,
    labels="inferred",
    label_mode="int",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

# Improve validation data loading
val_ds = val_ds.prefetch(AUTOTUNE)

print("\nClass Names:", classes)

# Data augmentation helps the model learn from different variations
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])

# Display sample images from the training dataset
plt.figure(figsize=(10, 6))

for i in range(min(8, len(X_train))):

    img = tf.keras.utils.load_img(
        X_train[i],
        target_size=IMG_SIZE
    )

    plt.subplot(2, 4, i + 1)
    plt.imshow(img)
    plt.title(classes[y_train[i]])
    plt.axis("off")

plt.suptitle("Sample Images from Cats vs Dogs Dataset")
plt.tight_layout()
plt.show()

print("\nCNN MODEL")

# Create the CNN model
model = models.Sequential([

    # Define the input image size
    layers.Input(shape=(128, 128, 3)),

    # Normalize pixel values from 0-255 to 0-1
    layers.Rescaling(1.0 / 255),

    # Apply image augmentation
    data_augmentation,

    # First convolution layer extracts basic image features
    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    # Reduce the spatial size of the feature map
    layers.MaxPooling2D((2, 2)),

    # Second convolution layer extracts more complex features
    layers.Conv2D(
        64,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    # Reduce feature map size
    layers.MaxPooling2D((2, 2)),

    # Third convolution layer extracts higher-level features
    layers.Conv2D(
        128,
        (3, 3),
        activation="relu",
        padding="same"
    ),

    # Reduce feature map size again
    layers.MaxPooling2D((2, 2)),

    # Convert feature maps into a one-dimensional vector
    layers.Flatten(),

    # Fully connected layer for classification
    layers.Dense(
        128,
        activation="relu"
    ),

    # Dropout helps reduce overfitting
    layers.Dropout(0.5),

    # Output layer with two classes: cat and dog
    layers.Dense(
        2,
        activation="softmax"
    )
])

# Display the CNN architecture
model.summary()

# Compile the model using Adam optimizer and cross-entropy loss
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\nModel compiled successfully!")

print("\nMODEL TRAINING")

# Set the number of training epochs
EPOCHS = 15

# Train the CNN model
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS
)

# Plot training and validation accuracy
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

# Plot training and validation loss
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid()
plt.show()

print("\nTESTING")

# Evaluate the model using the test dataset
test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

print("\nTest Loss:", round(test_loss, 4))
print("Test Accuracy:", round(test_accuracy * 100, 2), "%")

# Store actual and predicted labels
y_true = []
y_pred = []

# Generate predictions for all test images
for images, labels in test_ds:

    predictions = model.predict(
        images,
        verbose=0
    )

    # Select the class with the highest probability
    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    y_true.extend(labels.numpy())
    y_pred.extend(predicted_classes)

# Convert predictions into NumPy arrays
y_true = np.array(y_true)
y_pred = np.array(y_pred)

print("\nPredictions generated successfully.")
print("Total test images:", len(y_true))

print("Actual cats:", np.sum(y_true == 0))
print("Actual dogs:", np.sum(y_true == 1))
print("Predicted cats:", np.sum(y_pred == 0))
print("Predicted dogs:", np.sum(y_pred == 1))

# Display classification report
print("\nCLASSIFICATION REPORT")

print(
    classification_report(
        y_true,
        y_pred,
        labels=[0, 1],
        target_names=classes,
        zero_division=0
    )
)

# Create the confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1]
)

# Display confusion matrix
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=classes,
    yticklabels=classes
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix - Cats vs Dogs")
plt.show()

# Display sample predictions
plt.figure(figsize=(12, 8))

for images, labels in test_ds.take(1):

    predictions = model.predict(
        images,
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    for i in range(min(12, len(images))):

        plt.subplot(3, 4, i + 1)

        plt.imshow(
            images[i].numpy().astype("uint8")
        )

        actual = classes[labels[i].numpy()]
        predicted = classes[predicted_classes[i]]

        plt.title(
            f"Actual: {actual}\nPredicted: {predicted}"
        )

        plt.axis("off")

plt.suptitle("Sample Predictions")
plt.tight_layout()
plt.show()

# Find images that were classified incorrectly
misclassified_images = []
misclassified_actual = []
misclassified_predicted = []

for images, labels in test_ds:

    predictions = model.predict(
        images,
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    for i in range(len(images)):

        actual_label = labels[i].numpy()
        predicted_label = predicted_classes[i]

        if actual_label != predicted_label:

            misclassified_images.append(
                images[i].numpy()
            )

            misclassified_actual.append(
                classes[actual_label]
            )

            misclassified_predicted.append(
                classes[predicted_label]
            )

print("\nTotal Misclassified Images:",
      len(misclassified_images))

# Display misclassified images
num_images = min(
    9,
    len(misclassified_images)
)

if num_images > 0:

    plt.figure(figsize=(12, 8))

    for i in range(num_images):

        plt.subplot(3, 3, i + 1)

        plt.imshow(
            misclassified_images[i].astype("uint8")
        )

        plt.title(
            f"Actual: {misclassified_actual[i]}\n"
            f"Predicted: {misclassified_predicted[i]}"
        )

        plt.axis("off")

    plt.suptitle("Misclassified Images")
    plt.tight_layout()
    plt.show()

else:
    print("No misclassified images found.")

# Get the final training and validation results
train_accuracy = history.history["accuracy"][-1]
val_accuracy = history.history["val_accuracy"][-1]

train_loss = history.history["loss"][-1]
val_loss = history.history["val_loss"][-1]

# Calculate correct and incorrect predictions
correct_predictions = np.sum(
    y_true == y_pred
)

incorrect_predictions = np.sum(
    y_true != y_pred
)

# Display final performance summary
print("\nFINAL CNN PERFORMANCE SUMMARY")

print("Student Name          : SASTHA JEYASRI A")
print("Roll No               : 24BAD106")
print("Dataset               : Cats vs Dogs")
print("Classes               : Cat, Dog")

print("Training Accuracy     :",
      round(train_accuracy * 100, 2), "%")

print("Validation Accuracy   :",
      round(val_accuracy * 100, 2), "%")

print("Training Loss         :",
      round(train_loss, 4))

print("Validation Loss       :",
      round(val_loss, 4))

print("Testing Accuracy      :",
      round(test_accuracy * 100, 2), "%")

print("Total Test Images     :",
      len(y_true))

print("Correct Predictions   :",
      correct_predictions)

print("Incorrect Predictions :",
      incorrect_predictions)

# Save the trained CNN model
model.save(
    "/content/cats_vs_dogs_CNN.keras"
)

print("\nCNN model saved successfully!")
print("Location: /content/cats_vs_dogs_CNN.keras")
```
